* Now We have pandas dataframe where we can Now Perform a loop on the whole json dataset to convert them into pandas dataframe through a empty dataframe and a for loop *

```python
df=pd.DataFrame() #empty dataFrame
```

```python
df                   #printing empty data frame
```

```python
for i in range(1,501):
    response=requests.get('https://api.themoviedb.org/3/movie/top_rated?api_key=c294799cfbabd4c68b1295e1a0a5783f&language=en-US&page={}'.format(i))
    temp_df=pd.DataFrame(response.json()['results'])[['id','title','overview','release_date','popularity','vote_count','vote_average']]
    df=pd.concat([df,temp_df],ignore_index=True)
```
