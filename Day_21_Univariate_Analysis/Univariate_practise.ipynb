```python
import pandas as pd
import seaborn as sns
```

```python
df=pd.read_csv("train.csv")
```

```python
df
```

```python
df.head()
```

__1. Categorical Data__

a.Countplot 

```python
sns.countplot(x="Survived",data=df,palette=['Red','Green'])  
```

```python
sns.countplot(
    x="Survived",
    hue="Survived",
    data=df,
    palette=['Red', 'Green'],
    legend=False
)
```

```python
df['Survived'].value_counts()
```

```python
df['Survived'].value_counts().plot(kind='bar')
```

```python
sns.countplot(x='Pclass',hue='Pclass',data=df,palette=['Red','Green','Yellow'])  
```

```python
sns.countplot(
    x="Sex",
    hue="Sex",
    data=df,
    palette=['Red', 'Green'],
    legend=False
)
```

b. Pie Chart

```python
df['Survived'].value_counts().plot(kind='pie')
```

```python
df['Survived'].value_counts().plot(kind='pie',autopct='%.2f')
```

```python
df['Pclass'].value_counts().plot(kind='pie',autopct='%.2f')
```

```python
df['Sex'].value_counts().plot(kind='pie',autopct='%.2f')
```
